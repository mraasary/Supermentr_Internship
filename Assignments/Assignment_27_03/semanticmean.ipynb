```python
import spacy

# Load English model (run once in terminal: python -m spacy download en_core_web_sm)
nlp = spacy.load("en_core_web_sm")

# Word pairs
word_pairs = [
    ("happy", "joyful"),
    ("car", "vehicle"),
    ("teacher", "student"),
    ("hot", "cold"),
    ("dog", "animal")
]

# Calculate similarity
for w1, w2 in word_pairs:
    token1 = nlp(w1)
    token2 = nlp(w2)
    
    similarity = token1.similarity(token2)
    
    print(f"{w1} - {w2} : {round(similarity, 2)}")
```
