```python
from sklearn.feature_extraction.text import TfidfVectorizer

# List of documents
documents = [
    "Machine learning is amazing and powerful",
    "Artificial intelligence and machine learning are related",
    "Data science uses machine learning techniques",
    "Deep learning is a part of artificial intelligence",
    "Python is widely used in data science and AI"
]

# Create TF-IDF model
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

# Get feature names
features = vectorizer.get_feature_names_out()

# Convert matrix to array
tfidf_array = tfidf_matrix.toarray()

# Print top 3 keywords for each document
for i in range(len(documents)):
    print(f"\nDocument {i+1}:")
    scores = tfidf_array[i]
    top_indices = scores.argsort()[-3:]
    
    for idx in top_indices:
        print(features[idx], ":", round(scores[idx], 2))
```
