```python
# Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Step 1: Sample dataset (text + label)
data = {
    'message': [
        "Win money now", "Hello friend how are you",
        "Get free lottery tickets", "Let's meet tomorrow",
        "Claim your prize now", "Good morning",
        "Free entry in contest", "Call me when free"
    ],
    'label': [
        "spam", "ham", "spam", "ham",
        "spam", "ham", "spam", "ham"
    ]
}

df = pd.DataFrame(data)

# Step 2: Convert labels to numbers
df['label'] = df['label'].map({'ham': 0, 'spam': 1})

# Step 3: Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    df['message'], df['label'], test_size=0.2, random_state=42
)

# Step 4: Convert text to numerical features
vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Step 5: Train model
model = MultinomialNB()
model.fit(X_train_vec, y_train)

# Step 6: Test model
y_pred = model.predict(X_test_vec)
print("Accuracy:", accuracy_score(y_test, y_pred))

# Step 7: Predict new message
new_message = ["Congratulations! You won a free gift"]
new_message_vec = vectorizer.transform(new_message)

prediction = model.predict(new_message_vec)

if prediction[0] == 1:
    print("Spam Message")
else:
    print("Not Spam")
```

```python
# -------------------------------
# Spam Classifier Design (Thinking)
# -------------------------------

# 1. FEATURES USED IN SYSTEM
features = [
    "Presence of spam keywords (free, win, prize)",
    "Message length",
    "Use of capital letters",
    "Special characters (!, $, %)",
    "Presence of links (http, www)"
]

print("Features used for spam detection:")
for f in features:
    print("-", f)


# -------------------------------
# 2. DATA REQUIRED
# -------------------------------
data_required = {
    "messages": [
        "Win a free prize now",
        "Hello, how are you?",
        "Limited time offer!!!",
        "Let's meet tomorrow"
    ],
    "labels": ["spam", "ham", "spam", "ham"]
}

print("\nSample Data:")
for msg, label in zip(data_required["messages"], data_required["labels"]):
    print(f"{msg} --> {label}")


# -------------------------------
# 3. SIMPLE SPAM DETECTION LOGIC
# -------------------------------
def detect_spam(message):
    spam_words = ["free", "win", "offer", "prize"]

    message_lower = message.lower()

    for word in spam_words:
        if word in message_lower:
            return "Spam"

    return "Not Spam"


# Test the system
test_message = "Congratulations! You win a free gift"
result = detect_spam(test_message)

print("\nTest Message:", test_message)
print("Prediction:", result)


# -------------------------------
# 4. POSSIBLE MISTAKES
# -------------------------------
mistakes = [
    "Normal message may be classified as spam (False Positive)",
    "Spam message may not be detected (False Negative)",
    "New spam words not in system",
    "Different languages may cause errors"
]

print("\nPossible Mistakes:")
for m in mistakes:
    print("-", m)
```
