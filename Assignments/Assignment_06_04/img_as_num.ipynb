Description : Load an image, print shape, pixel values, channels, and explain them.

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# Load image
img = Image.open("image.jpg")
arr = np.array(img)

# 1. Shape
print("Shape:", arr.shape)        # (Height, Width, Channels)
print("dtype:", arr.dtype)        # uint8
print("ndim:", arr.ndim)          # 3
print("size:", arr.size)          # H x W x C

# 2. Pixel values
print("\nPixel at [0,0]:", arr[0, 0])          # [R, G, B]
print("First 3x3 pixels:\n", arr[:3, :3])
print("Min pixel value:", arr.min())
print("Max pixel value:", arr.max())

# 3. Channels
R = arr[:, :, 0]   # Red channel
G = arr[:, :, 1]   # Green channel
B = arr[:, :, 2]   # Blue channel

print("\nRed channel shape:", R.shape)
print("Green channel shape:", G.shape)
print("Blue channel shape:", B.shape)

print("\nRed avg:  ", round(R.mean(), 2))
print("Green avg:", round(G.mean(), 2))
print("Blue avg: ", round(B.mean(), 2))

# 4. Visualize channels
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

axes[0].imshow(arr)
axes[0].set_title("Original")
axes[0].axis("off")

axes[1].imshow(R, cmap="Reds")
axes[1].set_title("Red Channel")
axes[1].axis("off")

axes[2].imshow(G, cmap="Greens")
axes[2].set_title("Green Channel")
axes[2].axis("off")

axes[3].imshow(B, cmap="Blues")
axes[3].set_title("Blue Channel")
axes[3].axis("off")

plt.tight_layout()
plt.show()
```
